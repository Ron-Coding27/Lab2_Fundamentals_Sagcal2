```python
# [CELL 1] TASK 1
from main import surname, SEED_NUM
import random

print("Surname:", surname)
print("Seed Number:", SEED_NUM)

random.seed(hash(surname) + int(SEED_NUM))
readings = [random.randint(-10, 120) for _ in range(10)]

valid = [r for r in readings if 0 <= r <= 100]
invalid = [r for r in readings if r < 0 or r > 100]

def classify(r):
    if r <= 40:
        return "NORMAL"
    elif r <= 70:
        return "WARNING"
    else:
        return "CRITICAL"

classification = [classify(r) for r in valid]

print("Generated:", readings)
print("Valid:", valid)
print("Invalid:", invalid)
print("Classification:", classification)
print("Summary:", {
    "NORMAL": classification.count("NORMAL"),
    "WARNING": classification.count("WARNING"),
    "CRITICAL": classification.count("CRITICAL"),
    "FAULT": len(invalid)
})
```

```python
from main import surname, SEED_NUM

print("Surname:", surname)
print("Seed Number:", SEED_NUM)

signal = surname + str(int(SEED_NUM))

print("\nAssessment Data:")
print("Generated Signal:", signal)

processed = signal.upper().replace(" ", "")
print("Processed Signal:", processed)

letters = sum(c.isalpha() for c in processed)
digits = sum(c.isdigit() for c in processed)
spaces = sum(c.isspace() for c in processed)
symbols = sum(not c.isalnum() and not c.isspace() for c in processed)

print("Character Analysis:", {
    "Letters": letters,
    "Digits": digits,
    "Spaces": spaces,
    "Symbols": symbols
})

if letters > 0 and digits > 0:
    classification = "ALPHANUMERIC SIGNAL"
elif letters > 0:
    classification = "ALPHA SIGNAL"
elif digits > 0:
    classification = "NUMERIC SIGNAL"
else:
    classification = "UNKNOWN SIGNAL"

print("Signal Classification:", classification)

print("\nGenerated:", signal)
print("Processed:", processed)
print("Letters:", letters, "Digits:", digits, "Spaces:", spaces, "Symbols:", symbols)

print("\nClassification:", classification)
```

```python
from main import surname, SEED_NUM

print("Surname:", surname)
print("Seed Number:", SEED_NUM)

password = surname + str(int(SEED_NUM))

attempt_limit = int(SEED_NUM)

print("Generated Password:", password)
print("Attempt Limit:", attempt_limit)

attempts = 0
access_result = "Access Denied"
final_state = "System Locked"

test_attempts = ["wrong", "Sagcal", password]

for entry in test_attempts:
    attempts += 1
    if entry == password:
        access_result = "Access Granted"
        final_state = "System Unlocked"
        break
    if attempts >= attempt_limit:
        break

print("Attempts Made:", attempts)
print("Access Result:", access_result)
print("Final System State:", final_state)

print("\nExecution Log:")
print("Password:", password)
print("Attempt Limit:", attempt_limit)
print("Attempts Made:", attempts)
print("Result:", access_result)

print("\nSystem State:", final_state)
```
